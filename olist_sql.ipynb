```python
import sqlite3
import pandas as pd
```

```python
conn = sqlite3.connect("data/olist.db")

files = {
    "orders": "olist_orders_dataset.csv",
    "customers": "olist_customers_dataset.csv",
    "order_items": "olist_order_items_dataset.csv",
    "products": "olist_products_dataset.csv",
    "payments": "olist_order_payments_dataset.csv",
    "reviews": "olist_order_reviews_dataset.csv",
    "category": "product_category_name_translation.csv",
}
for name, f in files.items():
    pd.read_csv("data/" + f).to_sql(name, conn, if_exists="replace", index=False)
```

```python
# The SQL runs inside a triple-quoted string; pd.read_sql sends it to the database and returns a DataFrame
pd.read_sql("""
SELECT COUNT(*) AS n   -- count the rows and name the result column n
FROM orders            -- the table to count
""", conn)             # conn is the database connection created above
```
